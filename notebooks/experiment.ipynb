```python
targets = np.array(train_set.dataset.targets)[train_set.indices]
counts = np.bincount(targets, minlength=10)
plt.figure(figsize=(10,4))
plt.bar(CLASSES, counts, color='steelblue')
plt.title('Imágenes por clase en el subconjunto de entrenamiento')
plt.xticks(rotation=45); plt.ylabel('Cantidad'); plt.show()
print(dict(zip(CLASSES, counts)))
```
